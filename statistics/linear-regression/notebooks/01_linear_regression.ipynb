# Passo 1: Importar de bibliotecas

```python
import pandas as pd # Realiza operações de manipulação e análise de dados.
import statsmodels.api as sm # Estima o modelo de regressão linear via OLS.
import pymannkendall as mk # Teste de Mann-Kendall para detectar tendências em séries temporais.
```

# Passo 2: Ler o arquivo anual de CO₂

```python
# Leitura do arquivo CSV contendo os dados de CO₂. Representa a 
# média anual global de CO₂ em partes por milhão (ppm) desde 1979 até 2025.
df = pd.read_csv('../data/co2_annmean_gl.csv')

# Selecionar apenas as colunas "year" e "mean" para análise.
df = df[['year', 'mean']]

# Renomear "year" para "Ano" e "mean" para "CO2".
df = df.rename(columns={'year': 'Ano', 'mean': 'CO2'})

# Visualizar o DataFrame.
df
```

# Passo 3: Implementar o modelo de regressão linear simples via OLS

```python
# Foi criada uma variável chamada "Tempo" que representa o número de anos (de 1 a 47) 
# desde o início da série temporal. Isso é útil para a análise de regressão linear, 
# pois permite modelar a relação entre o tempo e os níveis de CO₂.

# Sem essa coluna (Tempo) é retornada a seguinte mensagem:

# [2] The condition number is large, 2.95e+05. This might indicate that there are
# strong multicollinearity or other numerical problems.

# Criando a coluna "Tempo".
df["Tempo"] = df["Ano"] - df["Ano"].min() + 1 

# Ajuste do modelo de regressão linear simples usando a fórmula 'CO2 ~ Tempo', 
# onde 'CO2' é a variável dependente e 'Tempo' é a variável independente.
modelo = sm.OLS.from_formula('CO2 ~ Tempo', data=df).fit()

# Imprime o resumo do modelo de regressão linear.
print(modelo.summary())
```

## Interpretação dos resultados da regressão linear simples

**1. O que a regressão mostra**

O modelo gerado foi:

$$
CO_2 = 330{,}5863 + 1{,}8895 \times Tempo
$$

ou

$$
Y = \beta_0 + \beta_1 X
$$

A concentração de CO₂ aumentou, em média, 1,89 ppm de CO₂ por ano entre 1979 e 2025.

O intervalo de confiança de 95% é:

$$
1{,}829 \leq \beta_1 \leq 1{,}950
$$

Portanto, a tendência estimada (1,8895) é bastante consistente.

O R²=0,989 também mostra que 98,9% da variabilidade do CO₂ é explicada pela tendência linear temporal.

**Houve aumentou ou redução na quantidade de CO₂ entre 1979 e 2025 pelo modelo de regressão linear simples?**

O número de anos entre 2025 e 1979 é:

`2025 − 1979 = 46 anos`

a variação estimada pelo modelo foi de:

`ΔCO₂ = 1,8895 × 46`

`ΔCO₂ ≈ 86,92 ppm`

Segundo a tendência linear, **houve aumento** de **86,92** ppm de CO₂ no período de 1979 a 2025.


**2. Magnitude da variação**

E sobre a magnitude da variação em ppm? Usam-se valores observados da série:
$$
ΔCO₂ = CO_{2,2025} − CO_{2,1979}
$$
$$
ΔCO₂ = 425{,}65 - 336{,}85
$$
$$
ΔCO₂ = 88,80
$$
e:

$$
\%\Delta CO_2 = \frac{CO_{2,2025} - CO_{2,1979}}{CO_{2,1979}} \times 100
= \frac{425{,}65 - 336{,}85}{336{,}85} \times 100
= 26{,}36\%
$$

Este valor informa que houve um aumento de **88,80 ppm** **(26,36%)** na concentração de CO₂ no período de 1979 a 2025.

**3) Limitação do modelo de regressão linear simples**

O teste de Durbin-Watson varia aproximadamente entre 0 e 4:

- próximo de 2 → pouca autocorrelação;
- próximo de 0 → forte autocorrelação positiva;
- próximo de 4 → forte autocorrelação negativa.

Neste caso, o resultado foi **DW = 0,050**, valor extremamente baixo, indicando **forte autocorrelação serial positiva** nos resíduos.

Isso significa que os erros não são independentes ao longo do tempo. Assim, embora a regressão linear continue útil para descrever a tendência média, os testes de significância e os intervalos de confiança do OLS **podem ficar excessivamente otimistas**.

Por esse motivo, foi realizada uma análise complementar com **Mann-Kendall** e com a **correção de Hamed-Rao**, que é mais apropriada diante da dependência temporal observada.

# Passo 4: Análise complementar a regressão linear simples

A pergunta científica:

**Qual é a tendência temporal das concentrações de CO₂ atmosférico entre 1979 e 2025? E qual é a magnitude de sua variação ao longo desse período?**

pode ser respondida com:

**Análise 1 — Regressão linear**

$$
CO_2 = β0 + β1 \times Tempo + ϵ
$$

$$
\beta_1 = 1,88 \, \text{ppm/ano}
$$


→ estima a taxa (ano) de aumento.

**Análise 2 — magnitude observada**

$$
ΔCO₂ = CO_{2,2025} − CO_{2,1979} = 425,65 - 336,85 = 88,80
$$

$$
ΔCO₂ = 88,80 \, \text{ppm}
$$
→ quantifica o aumento absoluto.

**Análise 3 — variação percentual observada**

$$
\%\Delta CO_2 = \frac{CO_{2,2025} - CO_{2,1979}}{CO_{2,1979}} \times 100
= \frac{425{,}65 - 336{,}85}{336{,}85} \times 100
= 26{,}36\%
$$

$$
\%\Delta CO_2 = 26{,}36\%
$$

→ mostra o aumento relativo.

**Análise 4 — Análise de tendência**

Calcula-se a **tendência** utilizando **Mann-Kendall** e **Sen's slope**.

Para uma série temporal é importante este cálculo porque é possível verificar se existe uma tendência monotônica e além disso estimar uma inclinação robusta, sem depender tanto dos pressupostos da regressão via OLS.

**O que esperamos responder?**

- Mann-Kendall → responde se existe uma tendência monotônica estatisticamente significativa;
- Sen's slope → estima a magnitude da tendência, de forma robusta aos outliers;
- OLS → fornece a tendência linear tradicional, que já foi calculada.

```python
# Aplicar Mann-Kendall original para testar a significância da tendência na série temporal de CO₂.
resultado_mk = mk.original_test(df["CO2"])

# Visualizar o resultado do teste de Mann-Kendall.
print(resultado_mk)
```

**Conclusão: Existe uma tendência monotônica crescente** e **estatisticamente significativa** (p < 0,05) na concentração de CO₂ entre 1979 e 2025.

Embora o resultado tenha sido `p = 0,0`, isso não significa que a probabilidade seja exatamente zero. Esse resultado indica que o p-valor foi tão pequeno que acabou sendo arredondado na apresentação. Portanto, a interpretação correta é que há evidência estatística muito forte de tendência, isto é, `p < 0,05`.

```python
# Resultado do Sen's slope. Ele representa a taxa de aumento do CO₂ ao longo do tempo.
print(f"Sen's slope: {resultado_mk.slope:.4f}")
```

Devido à forte autocorrelação serial positiva observada nos resíduos, foi aplicada a correção de **Hamed-Rao** ao teste de Mann-Kendall para avaliar a robustez da significância estatística da tendência.

Em síntese:

- Mann-Kendall original testa se existe tendência monotônica;
- Hamed-Rao ajusta a variância do teste para considerar a autocorrelação serial.

```python
# Garantir ordenação temporal.
df = df.sort_values("Ano").copy()

# Série temporal. Remove valores ausentes, se houver.
co2 = df["CO2"].dropna()

# ==========================================
# Mann-Kendall original.
# ==========================================

mk_original = mk.original_test(co2)

# ==========================================
# Mann-Kendall com correção para autocorrelação.
# ==========================================

# Testa a significância da tendência considerando a autocorrelação.
mk_hamed = mk.hamed_rao_modification_test(co2)

# ==========================================
# Resultados.
# ==========================================

print("\n==============================")
print("MANN-KENDALL ORIGINAL")
print("==============================")

print(f"Tendência: {mk_original.trend}")
print(f"p-valor:   {mk_original.p:.6f}")
print(f"Tau:       {mk_original.Tau:.4f}")
print(f"Sen slope: {mk_original.slope:.4f}")


print("\n==============================")
print("MANN-KENDALL HAMED-RAO")
print("==============================")

print(f"Tendência: {mk_hamed.trend}")
print(f"p-valor:   {mk_hamed.p:.6f}")
print(f"Tau:       {mk_hamed.Tau:.4f}")
print(f"Sen slope: {mk_hamed.slope:.4f}")
```

Os resultados confirmam tendência crescente pelos diferentes métodos, e a correção de Hamed-Rao fornece uma avaliação mais apropriada da significância na presença de autocorrelação.

| Método | Tendência | Magnitude | p-valor |
|---|---|---:|---:|
| OLS | Crescente | 1,89 ppm/ano | < 0,001* |
| Mann-Kendall | Crescente | — | < 0,001 |
| Sen's slope | Crescente | 1,89 ppm/ano | < 0,001 |

*Com ressalva da autocorrelação dos resíduos.

# Resultados consolidados

```python
co2_1979 = df.loc[df["Ano"] == 1979, "CO2"].iloc[0]
co2_2025 = df.loc[df["Ano"] == 2025, "CO2"].iloc[0]

variacao_abs = co2_2025 - co2_1979
variacao_pct = (variacao_abs / co2_1979) * 100

print(f"CO2 em 1979: {co2_1979:.2f}")
print(f"CO2 em 2025: {co2_2025:.2f}")
print(f"Variação absoluta: {variacao_abs:.2f}")
print(f"Variação percentual: {variacao_pct:.2f}%")
```

**Observação complementar**

Há uma pequena diferença entre a magnitude observada e a magnitude estimada pela tendência linear:

$$ 425{,}65\,\text{ppm} - 336{,}85\,\text{ppm} = \mathbf{88{,}80}\,\text{ppm} $$

$$ 1{,}8895\,\text{ppm/ano} \times 46\,\text{anos} = \mathbf{86{,}92}\,\text{ppm} $$

Isso mostra que o aumento observado foi ligeiramente superior em 2,16% ao estimado pela tendência linear média.

Assim, pode-se destacar:

- **88,80 ppm** e **26,36%** como a magnitude observada da variação;
- **1,89 ppm/ano** como a taxa média de tendência.
